How do we mechanically prove a theorem given only incidences?


```python
import numpy as np #import numpy
import json as js
```

```python
f=open("testfiles\Test1.cjs")
teststring = f.read()
```

```python
class Point:
    def __init__(self, name, homog):
        self.name = name
        self.homog = np.array(homog)
        self.x = self.homog[0]/self.homog[2]
        self.y = self.homog[1]/self.homog[2]

class Line:
    def __init__(self, name, homog):
        self.name = name
        self.homog = np.array(homog)

class Conic:
    def __init__(self, name, params):
        self.name = name
        self.params = np.array(params)


class Construction:
    def __init__(self, points, lines, conics, incidences):
        self.points = points
        self.lines = lines
        self.conics = conics
        self.incidences = incidences



def extract_information(test_string):
    x = teststring.replace("deselect();", "")
    x = x.replace("reset();", "")
    x = x.replace("update();", "")
    x = x.replace("recalcall();", "")
    x = x.replace(" ", "")
    x = x.replace("\n", "")
    x = x.replace("gslp=", "")
    x = x[0:-1]

    geom_array = js.loads(x)

    points = []
    lines = []
    conics = []
    incidences = {}
    name_dict = {}

    for geom_object in geom_array:
        if geom_object['type'] =='P':
            name = geom_object['name']
            homog = geom_object['pos']

            new_point = Point(name, homog)
            name_dict[name] = new_point
            points.append(new_point)

        if geom_object['type'] =='L':
            name = geom_object['name']
            homog = geom_object['pos']

            new_line = Line(name, homog)
            name_dict[name] = new_line
            lines.append(new_line)

        if geom_object['type'] =='C':
            name = geom_object['name']
            params = geom_object['pos']

            new_conic = Conic(name, params)
            name_dict[name] = new_conic
            conics.append(new_conic)

    for geom_object in geom_array:
        try:
            incids = geom_object["incidences"]
            incid_array = []
            for incid in incids:
                incid_array.append(name_dict[incid])
            incidences[name_dict[geom_object['name']]] = incid_array
        except:
            pass
        
    return Construction(points, lines, conics, incidences)

```

```python
constr = extract_information(teststring)
```
